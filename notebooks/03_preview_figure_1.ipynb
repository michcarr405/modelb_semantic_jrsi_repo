Figure 1. Inheritance fidelity stabilizes syntactic correlations, but adaptive gain requires sequence-selective interaction.
(a) Change in total motif–metabolite mutual information across inheritance fidelity in sequence-selective and sequence-agnostic regimes. (b) Adaptive gain across the same fidelity range. Syntactic MI increased significantly in both regimes, whereas adaptive gain was significant only in the sequence-selective regime at high inheritance fidelity. Shaded regions indicate approximate 95% confidence intervals across 20 replicate simulations; intervals in panel (a) are narrow and largely obscured by the plotted lines.

Syntactic motif–metabolite information accumulates in both sequence-selective and sequence-agnostic regimes.

The default Model B parameterization was analyzed in sequence-selective and sequence-agnostic regimes to test whether total motif–metabolite mutual information is sufficient to identify biologically meaningful organization. Changes in motif–metabolite MI increased with inheritance fidelity in both regimes (Fig. 1a). Sign-randomization permutation tests showed that ΔMI became significant in the sequence-selective condition from inheritance fidelity 0.2 onward after Benjamini–Hochberg correction, and in the sequence-agnostic condition from inheritance fidelity 0.3 onward. Thus, inheritance fidelity was sufficient to stabilize statistical correlations between molecular motifs and local metabolite configurations, even when intrinsic motif–metabolite affinity was removed.

The adaptive consequences of these correlations differed sharply between regimes (Fig. 1b). In the sequence-selective condition, adaptive gain became statistically robust only at high inheritance fidelity, with significant ΔFitness at inheritance probabilities 0.9 and 1.0 after FDR correction. The sequence-agnostic condition showed no FDR-significant adaptive gain at any inheritance fidelity, despite substantial increases in total MI at high fidelity. At inheritance fidelity 1.0, sequence-agnostic simulations produced a larger increase in syntactic MI than sequence-selective simulations, yet adaptive gain remained indistinguishable from zero.

These results show that total motif–metabolite MI measures syntactic correlation rather than semantic organization by itself. Sequence-agnostic positional structure can generate substantial inherited correlation, but adaptive gain requires sequence-selective molecular interaction. This decoupling motivates the central decomposition used below: total MI is treated as stored syntactic information, whereas semantic organization is identified as the subset of inherited motif–metabolite correlation that is sequence-selective, dynamically stabilized, and viability-relevant.
