Figure 3. Semantic-information estimates are robust across biological viability definitions.
(a) Semantic information across inheritance fidelity using three viability definitions: excess over sequence-agnostic control, mean future fitness, and threshold survival. (b) Corresponding syntactic–semantic gap. All three definitions produced similar semantic trajectories, with increasing semantic information at high inheritance fidelity and only modest widening of the gap. Shaded regions indicate bootstrap 95% intervals; in some regions these intervals are narrow and largely obscured by the plotted lines.

Alternative biological viability definitions produced nearly identical semantic-information trajectories (Fig. 3a). Excess over control, mean future fitness, and threshold survival all yielded semantic information near 0.21 bits at low inheritance fidelity, followed by gradual increases through intermediate fidelity and a sharp rise at high fidelity. By inheritance fidelity 1.0, all three definitions estimated approximately 0.41–0.44 bits of semantic information. The syntactic–semantic gap remained small across low-to-intermediate fidelities and widened only modestly at high fidelity (Fig. 3b), indicating that most sequence-selective syntactic information was viability-relevant under all three criteria. These results show that the semantic decomposition is robust to the choice of viability definition and is not an artifact of the excess-over-control metric alone.
