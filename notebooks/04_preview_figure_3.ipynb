Figure 2. Intervention-based decomposition separates syntactic correlation from semantic information.
(a) Syntactic and semantic information across inheritance fidelity in sequence-selective and sequence-agnostic regimes using excess over control as the viability definition. (b) Syntactic–semantic gap. (c) Semantic efficiency, defined as semantic information divided by syntactic information. (d) Value of information, defined as actual viability minus full-scrambled viability. Sequence-agnostic systems accumulated syntactic information but showed zero semantic information, zero semantic efficiency, and low value of information. Shaded regions indicate bootstrap 95% intervals; in some panels these intervals are narrow and largely obscured by the plotted lines.

To distinguish syntactic motif–metabolite correlation from viability-relevant semantic organization, the intervention-based decomposition used excess over the sequence-agnostic control as the biological viability definition. This analysis separated total motif–metabolite mutual information into syntactic information and the subset of that information required to preserve viability under coarse-graining interventions. In the sequence-selective regime, syntactic information increased with inheritance fidelity and was accompanied by substantial semantic information (Fig. 2a). Estimated semantic information rose from approximately 0.21 bits at low inheritance fidelity to 0.44 bits at inheritance fidelity 1.0. By contrast, sequence-agnostic simulations accumulated syntactic information with increasing inheritance fidelity, but estimated semantic information remained zero across the full parameter range.

This separation was statistically robust. Across all inheritance fidelities, intervention-point viability was significantly higher in the sequence-selective regime than in the sequence-agnostic regime under the excess-over-control viability definition (Mann–Whitney U tests, 504 intervention points per regime and fidelity; Benjamini–Hochberg corrected p FDR<10−41 for all comparisons). The direction of the effect was consistent across the full sweep, with sequence-selective median viability exceeding sequence-agnostic median viability at every inheritance fidelity.

The decomposition produced distinct syntactic–semantic gap profiles (Fig. 2b). In the sequence-selective regime, the gap between syntactic and semantic information remained small, increasing only from approximately 0.05 bits to 0.09 bits across the inheritance-fidelity sweep. This indicates that most preserved motif–metabolite correlation was viability-relevant under the excess-over-control criterion. In the sequence-agnostic regime, the gap increased sharply with inheritance fidelity, reaching approximately 0.48 bits at inheritance fidelity 1.0. Thus, inheritance fidelity stabilized sequence-agnostic motif–metabolite correlations, but those correlations were eliminated by the semantic decomposition because they did not contribute to excess viability.

Semantic efficiency further emphasized this separation (Fig. 2c). Sequence-selective systems maintained high semantic efficiency across the inheritance-fidelity sweep, with semantic information accounting for approximately 80–84% of syntactic information. Sequence-agnostic systems had zero semantic efficiency at all inheritance fidelities. The value of information showed the same pattern (Fig. 2d): sequence-selective systems exhibited strong positive viability value, increasing sharply at high inheritance fidelity, whereas sequence-agnostic systems remained near zero.

Together, these results show that sequence-agnostic structure can generate inherited syntactic information, but not viability-relevant semantic organization under the biological decomposition. Semantic information in Model B is therefore not equivalent to total mutual information. It is the subset of inherited motif–metabolite correlation that remains necessary for maintaining excess viability after sequence-agnostic structure is removed as a control.
