# Generate stats tables and manuscript figures


```python
from pathlib import Path
import sys

sys.path.insert(0, str((Path.cwd().parent / 'src').resolve()))

from modelb_semantic_repo.stats import build_baseline_stats, build_viability_gap_table, build_robustness_table, build_intervention_tests, build_method_ablation_tests, build_method_summary_table, build_viability_definition_tests
from modelb_semantic_repo.figures import make_all_figures
```

```python
mode = 'paper'
results_dir = Path('../results') / mode
stats_dir = Path('../stats') / mode
figures_dir = Path('../figures') / mode
stats_dir.mkdir(parents=True, exist_ok=True)
figures_dir.mkdir(parents=True, exist_ok=True)
```

```python
build_baseline_stats(results_dir).to_csv(stats_dir / 'baseline_permutation_tests.csv', index=False)
build_viability_gap_table(results_dir).to_csv(stats_dir / 'viability_gap_table.csv', index=False)
build_robustness_table(results_dir).to_csv(stats_dir / 'robustness_table.csv', index=False)
build_intervention_tests(results_dir).to_csv(stats_dir / 'intervention_selective_vs_control_tests_primary_no_kmeans.csv', index=False)
build_method_ablation_tests(results_dir).to_csv(stats_dir / 'method_ablation_kmeans_vs_primary_tests.csv', index=False)
build_method_summary_table(results_dir).to_csv(stats_dir / 'method_viability_summary.csv', index=False)
build_viability_definition_tests(results_dir).to_csv(stats_dir / 'viability_definition_dependence_tests.csv', index=False)
make_all_figures(results_dir, figures_dir)
figures_dir
```
