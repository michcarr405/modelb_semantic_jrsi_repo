\caption{\textbf{Method ablation supports exclusion of the k-means-profile diagnostic from the primary semantic frontier.}
(a) Semantic-information estimates across inheritance fidelity for the primary frontier excluding k-means interventions, the all-pooled diagnostic frontier, and the k-means-profile diagnostic. 
(b) Maximum frontier viability for the primary frontier and the k-means-profile diagnostic. 
Colours denote interaction regime, and line/marker styles denote method family. 
The primary frontier, with k-means interventions excluded, recovered increasing semantic information in the sequence-selective regime while sequence-agnostic controls remained near zero. 
The k-means-profile diagnostic preserved substantially greater maximum frontier viability in sequence-selective runs, particularly at high inheritance fidelity, whereas the corresponding sequence-agnostic controls remained near baseline. 
This selective-control asymmetry indicates that k-means grouping preserved functionally meaningful affinity-profile similarity specifically when motif identity carried sequence-selective interaction structure. 
The k-means-profile diagnostic is therefore biologically informative, but it tests a more specific mechanistic property---whether affinity-profile similarity itself preserves viability---rather than the amount of generic motif--metabolite information required to preserve viability. 
For this reason, the k-means-profile diagnostic was treated as a diagnostic analysis rather than included in the primary pooled semantic frontier. 
Excluding it keeps the primary semantic estimate conservative and conceptually clean while showing that affinity-profile similarity can itself preserve viability in the sequence-selective regime.}

A method-ablation analysis was used to assess whether the primary semantic estimates depended on the k-means-profile diagnostic intervention family (Fig.~S\ref{fig:method_ablation}). The primary frontier excluding k-means interventions recovered increasing semantic information with inheritance fidelity in the sequence-selective regime, whereas sequence-agnostic controls remained near zero. The k-means-profile diagnostic preserved substantially greater maximum frontier viability in sequence-selective runs, especially at high inheritance fidelity, but did not produce comparable viability preservation in sequence-agnostic controls. This asymmetry indicates that affinity-profile similarity is biologically meaningful in the sequence-selective regime, while supporting the decision to exclude the diagnostic from the primary pooled frontier to keep the main semantic estimate conservative and conceptually specific.
