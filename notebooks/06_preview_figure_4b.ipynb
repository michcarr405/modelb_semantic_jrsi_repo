\subsection{Generic entropy-state definitions do not identify semantic information}

Entropy-based state definitions were evaluated as an additional specificity test for the semantic decomposition. Four definitions were compared: negative fitness-distribution entropy, negative local-configuration entropy, negative adjacency-state entropy, and negative compositional entropy. In contrast to the biologically aligned viability definitions, all four entropy-state definitions yielded semantic information near zero across the full inheritance-fidelity sweep (Fig.~\ref{fig:entropy_state_definitions}a).

The corresponding syntactic--semantic gaps increased with inheritance fidelity under all four definitions (Fig.~\ref{fig:entropy_state_definitions}b). The gap rose from approximately $0.26$ bits at low inheritance fidelity to approximately $0.52$ bits at inheritance fidelity $1.0$, indicating that inherited motif--metabolite correlations were present but were not retained as semantic information under these generic entropy-state criteria.

These results provide a specificity control for the intervention-based decomposition. The decomposition does not automatically classify inherited syntactic correlation as semantic information under arbitrary viability definitions. Instead, nonzero semantic information is recovered only when the viability criterion is biologically aligned with the modelled organization of local metabolite configurations and adaptive consequences.
